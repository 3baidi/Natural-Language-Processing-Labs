```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


docs_task1 = [
    'This is the first document.',
    'This document is the second document.',
    'And this is the third one.',
    'Is this the first document?'
]


tfidf_vectorizer = TfidfVectorizer()
vector_matrix = tfidf_vectorizer.fit_transform(docs_task1)
cosine_sim_matrix = cosine_similarity(vector_matrix)


df_cosine = pd.DataFrame(data=cosine_sim_matrix, index=docs_task1, columns=docs_task1)
display(df_cosine)
```

```python

docs_task2 = [
    'data science is one of the most important fields of science',
    'this is one of the best data science courses',
    'data scientists analyze data'
]


tfidf_task2 = TfidfVectorizer()
matrix_task2 = tfidf_task2.fit_transform(docs_task2)


df_tfidf = pd.DataFrame(matrix_task2.toarray(), columns=tfidf_task2.get_feature_names_out())
display(df_tfidf)
```

```python
import pandas as pd
import re
import nltk
from nltk.tokenize import word_tokenize
import gensim

nltk.download('punkt')
nltk.download('punkt_tab')


df_simpsons = pd.read_csv('simpsons_script_lines.csv', engine='python', on_bad_lines='skip')
df_simpsons = df_simpsons.dropna(subset=['spoken_words'])


def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"[0-9]", '', text)
    text = re.sub(r"[)(,”“.’$-]", '', text)
    return text

df_simpsons['cleaned_words'] = df_simpsons['spoken_words'].apply(clean_text)


print("Tokenizing... (This may take a minute)")
tokens = [word_tokenize(sentence) for sentence in df_simpsons['cleaned_words']]


print("Training Word2Vec Model...")
Skip_gram_model = gensim.models.Word2Vec(tokens, min_count=1, vector_size=100, window=5, sg=1)
print("Model Trained Successfully!")
```

```python

print("Words similar to 'homer':")
print(Skip_gram_model.wv.most_similar('homer'))

print("\nWords similar to 'marge':")
print(Skip_gram_model.wv.most_similar('marge'))

print("\nWords similar to 'bart':")
print(Skip_gram_model.wv.most_similar('bart'))
```

```python

print("Odd one out of ['jimbo', 'milhouse', 'kearney']:")
print(Skip_gram_model.wv.doesnt_match(['jimbo', 'milhouse', 'kearney']))

print("\nOdd one out of ['nelson', 'bart', 'milhouse']:")
print(Skip_gram_model.wv.doesnt_match(['nelson', 'bart', 'milhouse']))

print("\nOdd one out of ['homer', 'patty', 'selma']:")
print(Skip_gram_model.wv.doesnt_match(['homer', 'patty', 'selma']))
```
