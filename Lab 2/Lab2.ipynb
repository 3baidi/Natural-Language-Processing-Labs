```python
import pandas as pd
import re
from collections import Counter
import kagglehub
import os

path = kagglehub.dataset_download("seriousran/appletwittersentimenttexts")
file_path = os.path.join(path, "apple-twitter-sentiment-texts.csv")
df = pd.read_csv(file_path)

all_text = " ".join(df['text'].dropna().astype(str).str.lower())
hashtags = re.findall(r"#\w+", all_text)

print(f"Total hashtags found: {len(hashtags)}")
print("\n--- Top 10 Hashtags ---")
for tag, count in Counter(hashtags).most_common(10):
    print(f"{tag}: {count}")
```

```python
import re
text = "This year is 2021"
pattern = re.compile(r"\d+")

print("Type of compiled pattern:", type(pattern))
print("Updated text:", pattern.sub("2022", text))
```

```python
import re
text = "a 11 b 2 3 c 4"

resulting_list = re.split(r"\s*\d+\s*", text)
resulting_list = [item for item in resulting_list if item]

print("Resulting list:", resulting_list)
```

```python
import spacy
import nltk
nltk.download('punkt_tab')

text = "I'm enjoying the NLP course!"
nlp = spacy.load("en_core_web_sm")

print("--- spaCy Tokenization ---")
for token in nlp(text):
    print(token.text)

print("\n--- NLTK Tokenization ---")
for token in nltk.word_tokenize(text):
    print(token)
```
